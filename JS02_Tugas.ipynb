```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# 1. Membaca dataset
df = pd.read_csv('Tugas.csv')

print(df.shape)
df.head()
```

```python
# 2. Memisahkan variabel yang digunakan dan tidak digunakan

X = df.drop(columns=['id', 'diagnosis'])
y = df['diagnosis']
```

```python
# 3. Encoding kolom diagnosis
encoder = LabelEncoder()

y = encoder.fit_transform(y)

print("\nHasil encoding diagnosis:")
print("B = 0")
print("M = 1")

print("\n10 data diagnosis setelah encoding:")
print(y[:10])
```

```python
# Cek nilai kosong
print("Jumlah nilai kosong:")
print(X.isnull().sum())

# Cek nilai infinite
import numpy as np

print("\nJumlah nilai infinite:")
print(np.isinf(X).sum())

# Cek tipe data
print("\nTipe data:")
print(X.dtypes)
```

```python
print("Kolom yang memiliki NaN:")
print(X.columns[X.isnull().any()].tolist())

print("\nKolom yang memiliki infinite:")
print(X.columns[np.isinf(X).any()].tolist())
```
