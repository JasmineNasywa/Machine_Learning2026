```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
import time

# 1. Load dataset MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalisasi data (0-255 → 0-1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encoding label
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Fungsi untuk uji coba model dengan berbagai parameter (Eksperimen Tugas)
def run_mnist_experiment(hidden_units_list, activation_func, model_name):
    print(f"\n==============================================")
    print(f"MENJALANKAN: {model_name}")
    print(f"Arsitektur Hidden Neurons: {hidden_units_list} | Aktivasi: {activation_func}")
    print(f"==============================================")
    
    # 2. Bangun model JST
    model = Sequential([
        Flatten(input_shape=(28, 28)), # Ubah gambar 28x28 menjadi vektor
    ])
    
    # Tambah hidden layers dinamis sesuai parameter
    for units in hidden_units_list:
        model.add(Dense(units, activation=activation_func))
        
    # Output layer (10 kelas untuk angka 0-9)
    model.add(Dense(10, activation='softmax'))
    
    # 3. Kompilasi model
    model.compile(optimizer='adam', 
                  loss='categorical_crossentropy', 
                  metrics=['accuracy'])
    
    # 4. Latih model (catat waktu pelatihan)
    start_time = time.time()
    history = model.fit(X_train, y_train, epochs=5, batch_size=64, validation_split=0.1, verbose=1)
    end_time = time.time()
    
    training_time = end_time - start_time
    print(f" Waktu Pelatihan: {training_time:.2f} detik")
    
    # 5. Evaluasi model
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print(f" Akurasi pada Data Uji: {acc * 100:.2f}% (Loss: {loss:.4f})")
    
    return training_time, acc

# ==========================================
# EKSEKUSI EKSPERIMEN
# ==========================================

# 1. Baseline / Standar (Misal: 128 & 64 neuron dengan ReLU)
run_mnist_experiment([128, 64], 'relu', "Model Baseline [128, 64] - ReLU")

# 2. Ubah jumlah neuron jadi lebih besar (256 dan 128) sesuai instruksi tugas[cite: 6]
run_mnist_experiment([256, 128], 'relu', "Eksperimen 1: Neuron Lebih Besar [256, 128] - ReLU")

# 3. Tambah satu hidden layer lagi (jadi 3 hidden layers: 256, 128, 64)[cite: 6]
run_mnist_experiment([256, 128, 64], 'relu', "Eksperimen 2: Tambah 1 Hidden Layer [256, 128, 64] - ReLU")

# 4. Eksperimen fungsi aktivasi: Sigmoid vs ReLU (pada struktur [256, 128])[cite: 6]
run_mnist_experiment([256, 128], 'sigmoid', "Eksperimen 3: Menggunakan Aktivasi Sigmoid")
```
