```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('jakarta_house.csv')

# Tampilkan 5 data teratas
df.head()
```

```python
#inspeksi data

df.shape
df.info()
df.isnull().sum()
```

```python
#data amputation
# data imputation

# land_area - mean
df['land_area'].fillna(value=df['land_area'].mean(), inplace=True)

# building_area - modus
df['building_area'].fillna(value=df['building_area'].mode()[0], inplace=True)
```

```python
df.to_csv('jakarta_house_imputed.csv', index=False)
```
