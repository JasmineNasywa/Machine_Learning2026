## Kegunaan Library

1. PyPREP
PyPREP digunakan untuk melakukan preprocessing pada data EEG
(Electroencephalography). Library ini merupakan implementasi Python
dari PREP pipeline yang membantu proses preprocessing data EEG.

2. SciPy
SciPy merupakan library Python untuk komputasi ilmiah. Library ini
menyediakan berbagai fungsi untuk matematika, statistika, optimasi,
pemrosesan sinyal, dan komputasi numerik.

3. Weights & Biases (wandb)
wandb digunakan untuk membantu proses pelacakan dan pengelolaan
eksperimen machine learning, seperti mencatat parameter, metrik,
hasil eksperimen, dan perkembangan training model.

4. pyECG
pyECG merupakan library Python yang digunakan untuk membaca dan
menulis beberapa format dataset ECG (Electrocardiography), termasuk
format WFDB dan iShine.
