```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
#1. MEMBACA DATASET

df = pd.read_csv('tugas.csv')

print("Ukuran dataset:")
print(df.shape)

print("\nNama kolom:")
print(df.columns.tolist())
```

```python
# 2. MEMISAHKAN VARIABEL YANG DIGUNAKAN DAN TIDAK DIGUNAKAN

X = df.drop(columns=['id', 'diagnosis', 'Unnamed: 32'])
y = df['diagnosis']

print("\nJumlah fitur yang digunakan:", X.shape[1])
print("Jumlah data:", X.shape[0])

print("\nFitur yang digunakan:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
# 3. ENCODING KOLOM DIAGNOSIS

encoder = LabelEncoder()

y = encoder.fit_transform(y)

print("\nHasil encoding diagnosis:")
print("B = 0")
print("M = 1")

print("\n10 data diagnosis setelah encoding:")
print(y[:10])
```

```python
# 4. STANDARDISASI SEMUA KOLOM NUMERIK

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# Mengubah hasil standardisasi menjadi DataFrame
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

print("\nHasil standardisasi:")
print(X_scaled.head())
```

```python
# 5. STRATIFIED SPLIT DATA 80:20

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("\nHasil pembagian data:")
print("Data latih :", X_train.shape)
print("Data uji   :", X_test.shape)

print("\nJumlah data y_train:", len(y_train))
print("Jumlah data y_test :", len(y_test))
```

```python
# 6. CEK DISTRIBUSI KELAS SETELAH STRATIFIED SPLIT

print("\nDistribusi kelas data latih:")
print(pd.Series(y_train).value_counts())

print("\nDistribusi kelas data uji:")
print(pd.Series(y_test).value_counts())
```
