```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
dpath = 'jakarta_house_imputed.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
#slice data
df = df[['price', 'district', 'city', 'bed_rooms', 'bath_rooms', 'carport', 'land_area', 'building_area']]
df.head()
```

```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['district'] = le.fit_transform(df['district']) # proses encoding
df['city'] = le.fit_transform(df['city']) # proses encoding
```

```python
df.head()
```

```python
#standarisasi

std = StandardScaler()
df['price'] = std.fit_transform(df[['price']])
```
